```python
classification_features = [
    "Age",
    "Gender",
    "Class",
    "Study Time",
    "Absences",
    "Parental_Education",
    "Internet_Access",
    "Extracurricular_Activities",
    "Previous_Year_Score"
]

X_cls = df[classification_features]
y_cls = df["Pass_Fail"]

print("Features shape:", X_cls.shape)
print("\nTarget distribution:")
print(y_cls.value_counts())
```

```python
from sklearn.model_selection import train_test_split

X_train_cls, X_test_cls, y_train_cls, y_test_cls = train_test_split(
    X_cls,
    y_cls,
    test_size=0.20,
    random_state=42,
    stratify=y_cls
)

print("Training features:", X_train_cls.shape)
print("Testing features:", X_test_cls.shape)

print("\nTraining target:")
print(y_train_cls.value_counts())

print("\nTesting target:")
print(y_test_cls.value_counts())
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

numeric_cls = [
    "Age",
    "Class",
    "Study Time",
    "Absences",
    "Previous_Year_Score"
]

categorical_cls = [
    "Gender",
    "Parental_Education",
    "Internet_Access",
    "Extracurricular_Activities"
]

cls_preprocessor = ColumnTransformer(
    transformers=[
        (
            "categorical",
            OneHotEncoder(handle_unknown="ignore"),
            categorical_cls
        ),
        (
            "numeric",
            "passthrough",
            numeric_cls
        )
    ]
)

print("Classification preprocessor created successfully.")
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline

logistic_model = Pipeline(
    steps=[
        ("preprocessor", cls_preprocessor),
        ("model", LogisticRegression(max_iter=1000))
    ]
)

print("Logistic Regression model created successfully.")
```

```python
logistic_model.fit(X_train_cls, y_train_cls)

print("Logistic Regression model trained successfully.")
```

```python
logistic_pred = logistic_model.predict(X_test_cls)

print("First 20 predictions:")
print(logistic_pred[:20])
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

logistic_accuracy = accuracy_score(y_test_cls, logistic_pred)

logistic_precision = precision_score(
    y_test_cls,
    logistic_pred,
    pos_label="Fail"
)

logistic_recall = recall_score(
    y_test_cls,
    logistic_pred,
    pos_label="Fail"
)

logistic_f1 = f1_score(
    y_test_cls,
    logistic_pred,
    pos_label="Fail"
)

print("Logistic Regression Results")
print("---------------------------")
print("Accuracy :", round(logistic_accuracy, 3))
print("Precision:", round(logistic_precision, 3))
print("Recall   :", round(logistic_recall, 3))
print("F1 Score :", round(logistic_f1, 3))
```

```python
from sklearn.cluster import KMeans

kmeans_3 = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)

cluster_labels_3 = kmeans_3.fit_predict(X_cluster)

print("K-Means clustering completed.")
print("Number of clusters:", len(set(cluster_labels_3)))
```

```python
df["Cluster"] = cluster_labels_3

print("Cluster column added successfully.")
print("\nCluster distribution:")
print(df["Cluster"].value_counts().sort_index())
```

```python

from sklearn.metrics import silhouette_score

silhouette_results = []

for k in [3, 4, 5]:
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )
    
    labels = kmeans.fit_predict(X_cluster)
    
    score = silhouette_score(
        X_cluster,
        labels
    )
    
    silhouette_results.append({
        "K": k,
        "Silhouette Score": score
    })

silhouette_results = pd.DataFrame(silhouette_results)

print(silhouette_results.round(3))
```

```python
def student_estimation_panel(
    age,
    gender,
    student_class,
    study_time,
    absences,
    parental_education,
    internet_access,
    extracurricular_activities,
    previous_year_score
):
    
    student = pd.DataFrame({
        "Age": [age],
        "Gender": [gender],
        "Class": [student_class],
        "Study Time": [study_time],
        "Absences": [absences],
        "Parental_Education": [parental_education],
        "Internet_Access": [internet_access],
        "Extracurricular_Activities": [extracurricular_activities],
        "Previous_Year_Score": [previous_year_score]
    })

    # Predicted final percentage
    predicted_grade = ridge_model.predict(student)[0]

    # Pass/Fail prediction
    predicted_result = logistic_model.predict(student)[0]

    # Class probabilities
    probabilities = logistic_model.predict_proba(student)[0]
    classes = logistic_model.named_steps["model"].classes_

    probability_dict = dict(zip(classes, probabilities))

    print("=" * 45)
    print("       STUDENT ESTIMATION PANEL")
    print("=" * 45)

    print(f"Predicted Final Percentage : {predicted_grade:.2f}%")
    print(f"Predicted Result           : {predicted_result}")
    print(f"Fail Probability           : {probability_dict['Fail'] * 100:.2f}%")
    print(f"Pass Probability           : {probability_dict['Pass'] * 100:.2f}%")

    print("=" * 45)
```
